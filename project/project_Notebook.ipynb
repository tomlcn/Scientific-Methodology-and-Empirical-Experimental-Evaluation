## I - Introduction

Quick sort is a sorting algorithm and is based on partitioning of array of data into smaller arrays. A large array is partitioned into two arrays one of which holds values smaller than the specified value, say pivot, based on which the partition is made and another array holds values greater than the pivot value.
Quicksort partitions an array and then calls itself recursively twice to sort the two resulting subarrays. This algorithm is quite efficient for large-sized data sets as its average and worst-case complexity are O(n2), respectively.

Threads are a generic operating system concept representing a single flow of execution within a process, whereas POSIX threads (commonly known as pthreads) are a specific, standardized C-language API defined by the IEEE POSIX 1003.1c standard for creating and managing these threads on Unix-like systems. 
While "thread" is the fundamental unit of execution that shares the process's memory space and resources, pthreads provide the concrete programming interface (such as pthread_create and pthread_join) used to implement this concept in C and C++

In this project, we aim to compare 3 different thread configurations using a Quicksort algorithm. We have the following configurations:

- Threads working in parallel
- Threads working sequentially
- POSIX threads working sequentially

Thus, we expect the parallel configuration to be the worst of all three when the array size is small, then slowly reduce the gap before becoming better than both sequential configurations.

We can explain this by the fact that some time is needed to initialize these threads. For example, to process a small instruction set, the sequential algorithm can directly start working and finish while the parallel version is still preparing the threads. However, at a certain point, when we increase the problem size, the time needed for thread initialization will be compensated by the time saved through parallelization.

## II - Methodology

To evaluate and compare the performance of the sequential and parallel Quicksort implementations, we established a benchmarking methodology designed to test different array sizes.

Array Size Scaling: The performance tests were conducted across a broad range of input sizes, starting from small arrays (size 1) up to large scales (size $30,000,000$). This wide spectrum allows us to observe how each configuration execution time behave depending of problem size.

Statistical Analysis & Visualization: To ensure high statistical rigor, multiple repetitions were performed for each array size. Using an automated R pipeline, we computed the 95% confidence intervals and generated comparative log-log curves to clearly illustrate performance trends and isolate the exact crossover point. Furthermore, we conducted an Anova test to verify the effect of the size problem, the configuration type and the interaction between both.

The experimental results reveal distinct performance hierarchies depending on the size of the workload.

Small Array Sizes: For smaller datasets, the sequential implementation significantly outperforms the parallel configuration. The built-in C library qsort achieves the fastest execution times, followed by the sequential Quicksort, while the parallel Pthreads implementation have the highest execution latency.

Large Array Sizes: As the array size scales up, the performance hierarchy shifts. Tthe parallel Quicksort surpasses both sequential version, becoming faster as the workload increases.

Scalability Gap: The performance divergence between the parallel and sequential implementations growth proportionally as the array size grows, confirming that execution gains scale with problem magnitude. Log-log scale visualizations consistently capture this transition across independent test runs.

Statistical Validation: A two-way ANOVA test was conducted to evaluate the effects of algorithm type, array size, and their interaction on execution time. The analysis reveal effects for both algorithm type ($F = 30452, p < 2\times 10^{-16}$) and array size ($F = 47876, p < 2\times 10^{-16}$). Furthermore, the interaction between the algorithm type and the array size is statistically significant ($F = 16382, p < 2\times 10^{-16}$).

## IV - Discussion

Performance Inefficiency on Small Arrays: For smaller size, the sequential configuration outperforms the parallel version. This happend because the overhead associated with thread management—such as thread creation, synchronization outweighs the actual sorting workload. In this regime, the time spent preparing and launching parallel tasks exceeds the time required to sort the array sequentially.

The Crossover Point and Parallel Benefits: As the array size increases transforming the workload into a compute-bound problem. At a certain threshold (when size is between 100k-1M), the time saved by dividing the recursion  across multiple CPU cores compensates for time needed to the initial thread creation overhead. Consequently, the parallel implementation surpasses the sequential version on larger arrays.

Interpretation of Statistical Findings: The two-way ANOVA test strongly reinforces these physical observations. The highly significant main effects and interaction terms allow us to reject the null hypothesis, confirming that both array size and thread configuration have a statistically meaningful impact on execution time. It's important to recall that ANOVA does not "prove" a hypothesis in an absolute sense, it provides robust quantitative evidence that these factors and their interaction significantly shape performance outcomes.
